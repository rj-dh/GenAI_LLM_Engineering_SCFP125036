```python
from groq import Groq
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
```

```python
client = Groq(api_key=GROQ_API_KEY)
```

```python
messages=[
    {
        "role": "user",
        "content": "Explain ONE PIECE anime",
    }
]
```

## Input Prompt

```python
chat_completion = client.chat.completions.create(messages=messages, model="openai/gpt-oss-20b")
```

```python
print(chat_completion.choices[0].message.content)
```

```python
def ask_llm(user_query):
    messages=[
        {
            "role": "user",
            "content": user_query,
        }
    ]
    chat_completion = client.chat.completions.create(messages=messages, model="openai/gpt-oss-20b")
    return print(chat_completion.choices[0].message.content)
```

```python
history = []

while True:
    user_input = input("Bol bantai kya mangta:")

    if user_input.lower() == "exit":
        break

    messages={
        "role": "user",
        "content": user_input,
    }

    history.append(messages)
    chat_completion = client.chat.completions.create(messages=history, model="openai/gpt-oss-20b")
    response = chat_completion.choices[0].message.content
    print(response)

    ai_response = {
        "role": "assistant",
        "content": response,
    }
    history.append(ai_response)
```
